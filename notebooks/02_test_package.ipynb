```python
import os
import scanpy as sc
from scpipe.qc import compute_qc_metrics, is_outlier , flag_low_quality_cells,filter_cells_and_genes
from scpipe.preprocess import remove_doublets,normalisation_and_log_transformation,feature_selection
```

```python
if os.path.basename(os.getcwd()) == "notebooks":
    os.chdir("..")
```

```python
adata = sc.read_h5ad("data/pbmc3k_raw.h5ad")
compute_qc_metrics(adata)
qc_fail = flag_low_quality_cells(adata,5,3)
adata_clean = filter_cells_and_genes(adata, qc_fail)
print(qc_fail.sum(), adata_clean.shape)    # expected: 266 (2434, 13479)
```

```python
adata_clean = remove_doublets(adata_clean)
print(adata_clean.shape)                                   # expected: (2395, 13479)

           # expected: 2000
normalisation_and_log_transformation(adata_clean)
print(adata_clean.X.sum(axis=1)[:5])                       # expected: ~900 each

feature_selection(adata_clean)
print(adata_clean.var["highly_variable"].sum()) 
```
